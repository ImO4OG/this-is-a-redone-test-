qn.1..write a python program that creates a student class containing a student name and age .add a method called introduce()that dispalys the students information

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Student Name: {self.name}")
        print(f"Age: {self.age}")


# Create a student object
student1 = Student("Imran", 20)

# Call the introduce method
student1.introduce()
```

write a python program that creates a phone class.each phone should have a brand and storage capacity.add methods taht dispaly phone information and stimulate making a call .create two phone object and call their methods.

```python
class Phone:
    def __init__(self, brand, storage_capacity):
        self.brand = brand
        self.storage_capacity = storage_capacity

    def display_info(self):
        print(f"Brand: {self.brand}")
        print(f"Storage Capacity: {self.storage_capacity} GB")

    def make_call(self, number):
        print(f"{self.brand} is calling {number}...")


# Create two Phone objects
phone1 = Phone("Samsung", 128)
phone2 = Phone("iPhone", 256)

# Call methods for phone1
print("Phone 1:")
phone1.display_info()
phone1.make_call("0700123456")

print()

# Call methods for phone2
print("Phone 2:")
phone2.display_info()
phone2.make_call("0750987654")
```
